Probability Part - 1

1.Import libraries and create dataset


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Create employee dataset
data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110,
                    111, 112, 113, 114, 115],

    "Department": ["IT", "HR", "IT", "Finance", "HR",
                   "IT", "Finance", "IT", "HR", "Finance",
                   "IT", "HR", "Finance", "IT", "HR"],

    "Age": [22, 25, 28, 30, 24, 35, 40, 27, 29, 45,
            32, 26, 38, 31, 50],

    "Experience": [1, 2, 4, 6, 2, 10, 15, 3, 5, 18,
                   8, 3, 12, 7, 20],

    "Salary": [25000, 28000, 32000, 40000, 27000,
               55000, 80000, 35000, 38000, 95000,
               60000, 30000, 75000, 50000, 150000],

    "Performance_Score": [65, 70, 75, 80, 68,
                          85, 90, 78, 72, 95,
                          88, 74, 92, 82, 96],

    "Training_Completed": ["Yes", "No", "Yes", "Yes", "No",
                           "Yes", "Yes", "No", "Yes", "Yes",
                           "Yes", "No", "Yes", "Yes", "No"]
}

df = pd.DataFrame(data)

print(df)
```

2. Part A – Understanding Data Types

```python
# First 5 records
print("First 5 records:")
print(df.head())

# Last 5 records
print("\nLast 5 records:")
print(df.tail())

# Number of rows and columns
print("\nShape:")
print(df.shape)

# Column names
print("\nColumn names:")
print(df.columns)

# Data types
print("\nData types:")
print(df.dtypes)
```

3. Part B – Basic Data Exploration

```python
# Descriptive statistics
print("Descriptive Statistics:")
print(df.describe())

# Employees in each department
print("\nEmployees in each department:")
print(df["Department"].value_counts())

# Training completed
print("\nTraining completed:")
print(df["Training_Completed"].value_counts())

# Missing values
print("\nMissing values:")
print(df.isnull().sum())
```

4. Part C – Mean, Median and Mode

```python
columns = ["Age", "Experience", "Salary", "Performance_Score"]

for column in columns:
    print("\n", column)
    print("Mean:", df[column].mean())
    print("Median:", df[column].median())
    print("Mode:", df[column].mode().tolist())
```

5. Part D – Measures of Spread

```python
for column in columns:

    minimum = df[column].min()
    maximum = df[column].max()
    range_value = maximum - minimum
    variance = df[column].var()
    std_dev = df[column].std()
    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3 - q1

    print("\n", column)
    print("Minimum:", minimum)
    print("Maximum:", maximum)
    print("Range:", range_value)
    print("Variance:", variance)
    print("Standard Deviation:", std_dev)
    print("Q1:", q1)
    print("Q3:", q3)
    print("IQR:", iqr)
```

6. Part E – Skewness

```python
print("Skewness:")

for column in columns:
    print(column, ":", df[column].skew())
```

7. Part F – Histogram

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Salary"], bins=6, edgecolor="black")

plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.title("Salary Distribution")

plt.show()
```

8. Box Plot

```python
plt.figure(figsize=(6, 5))

plt.boxplot(df["Salary"])

plt.ylabel("Salary")
plt.title("Salary Box Plot")

plt.show()
```

9. Part G – Combined Statistical Summary

```python
summary = pd.DataFrame({
    "Mean": [
        df["Age"].mean(),
        df["Experience"].mean(),
        df["Salary"].mean(),
        df["Performance_Score"].mean()
    ],

    "Median": [
        df["Age"].median(),
        df["Experience"].median(),
        df["Salary"].median(),
        df["Performance_Score"].median()
    ],

    "Std_Deviation": [
        df["Age"].std(),
        df["Experience"].std(),
        df["Salary"].std(),
        df["Performance_Score"].std()
    ],

    "Skewness": [
        df["Age"].skew(),
        df["Experience"].skew(),
        df["Salary"].skew(),
        df["Performance_Score"].skew()
    ]
})

summary.index = [
    "Age",
    "Experience",
    "Salary",
    "Performance Score"
]

print(summary)
```

10. Part H – Remove ₹150,000 Salary

```python
# Remove employee with salary 150000
df_without_high_salary = df[df["Salary"] != 150000]

# Calculate new values
new_mean = df_without_high_salary["Salary"].mean()
new_median = df_without_high_salary["Salary"].median()
new_std = df_without_high_salary["Salary"].std()
new_skew = df_without_high_salary["Salary"].skew()

print("After removing ₹150,000 salary:")

print("Mean:", new_mean)
print("Median:", new_median)
print("Standard Deviation:", new_std)
print("Skewness:", new_skew)
```

11. Part I – Department Analysis

```python
# Mean salary
print("Mean Salary:")
print(df.groupby("Department")["Salary"].mean())

# Median salary
print("\nMedian Salary:")
print(df.groupby("Department")["Salary"].median())

# Standard deviation
print("\nSalary Standard Deviation:")
print(df.groupby("Department")["Salary"].std())
```

12. Part J – Final Challenge

```python
# Select numerical columns
numeric_columns = df.select_dtypes(include=np.number).columns

# Calculate statistics
for column in numeric_columns:

    print("\n==============================")
    print("Column:", column)
    print("==============================")

    print("Mean:", df[column].mean())
    print("Median:", df[column].median())
    print("Mode:", df[column].mode().tolist())
    print("Minimum:", df[column].min())
    print("Maximum:", df[column].max())

    print("Range:",
          df[column].max() - df[column].min())

    print("Variance:", df[column].var())
    print("Standard Deviation:", df[column].std())

    print("Q1:", df[column].quantile(0.25))
    print("Q3:", df[column].quantile(0.75))

    print("IQR:",
          df[column].quantile(0.75) -
          df[column].quantile(0.25))

    print("Skewness:", df[column].skew())
```
