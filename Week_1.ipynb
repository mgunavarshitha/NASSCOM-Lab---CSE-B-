# FUNCTIONS

```python
# 1.defining a basic functions
def greet():
    print("Hello! Welcome to Python Functions.")


greet()
```

```python
# 2. FUNCTION WITH PARAMETERS

def greet_user(name):
    print(f"Hello, {name}!")

greet_user("Alice")
greet_user("Bob")
```

```python
# 3.function with default paramters

def greet_user(name="Guest"):
  print(f"Welcome, {name}!")

greet_user()         # Uses default
greet_user("Akshaya")   # Overrides default
```

```python
#4. USING *args (VARIABLE NUMBER OF POSITIONAL ARGUMENTS)

def print_names(*names):
    print("Names passed:", names)
    for name in names:
        print("Hello", name)

print_names("Alice", "bob", "Charlie")
```

```python
#  6. USING **kwargs (VARIABLE NUMBER OF KEYWORD ARGUMENTS)

def print_info(**info):
    print("User Info:")
    for key, value in info.items():
        print(f"{key} = {value}")

print_info(name="Alice", age=25, city="Delhi")
```

```python
 # 7. COMBINING *args AND **kwargs

def show_details(*args, **kwargs):
    print("Args:", args)
    print("Kwargs:", kwargs)

show_details(1, 2, 3, name="Ravi", city="Mumbai")
```

```python
#  8. FUNCTION NESTING (CALL ONE FUNCTION INSIDE ANOTHER)

def multiply(a, b):
    return a * b

def square(n):
    return multiply(n, n)

print("Square of 4:", square(4))
```

```python
def student_info(**info):
  print("student info:")
  for key,value in info.items():
      print(f"{key}={value}")
student_info(name = "varsha",rollno=89,branch="CSE")
```

```python
numbers = list(map(int, input("Enter numbers separated by spaces: ").split()))

square= [x ** 2 for x in numbers]

even_squared = [x for x in squared if x % 2 == 0]

print("Squared elements:", squared)
print("Even squared elements:", even_squared)
```

# Python Practical: Lists, Sets, Tuples, Dictionaries

```python

fruits = ['apple', 'banana', 'cherry', 'apple']
print("Original list:", fruits)

print("First fruit:", fruits[0])

```

```python
fruits.append('orange')
print("After appending:", fruits)
```

```python
fruits.remove('banana')
print("After removing banana:", fruits)

print("Number of fruits:", len(fruits))
```

```python

dimensions = (10, 20, 30)
print("Tuple:", dimensions)
```

```python
print("Second element:", dimensions[1])
```

```python

unique_numbers = {1, 2, 3, 2, 1}
print("Set (no duplicates):", unique_numbers)


```

```python
unique_numbers.add(4)
print("After adding 4:", unique_numbers)
```

```python
unique_numbers.remove(2)
print("After removing 2:", unique_numbers)
```

```python
set_a = {1, 2, 3}
set_b = {3, 4, 5}
print("Union:", set_a | set_b)
print("Intersection:", set_a & set_b)
print("Difference (A - B):", set_a - set_b)
```

```python

student = {
    'name': 'Alice',
    'age': 21,
    'course': 'Python'
}
print("Dictionary:", student)


```

```python
student['grade'] = 'A'
print("After adding grade:", student)

del student['age']
print("After deleting age:", student)

for key, value in student.items():
    print(f"{key}: {value}")
```
