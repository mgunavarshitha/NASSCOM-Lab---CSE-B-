```python
X = df[["Study_Hours", "Attendance", "Assignment_Score"]].values
Y = df["Final_Score"].values

# Append column of ones for intercept tracking (b0)
X_model = np.column_stack((np.ones(len(X)), X))

# Estimate coefficients: beta = (X^T * X)^(-1) * X^T * Y
beta = np.linalg.inv(X_model.T @ X_model) @ X_model.T @ Y

print("Estimated Model Coefficients:")
print(f"b0 (Intercept): {beta[0]:.4f}")
print(f"b1 (Study Hours Coefficient): {beta[1]:.4f}")
print(f"b2 (Attendance Coefficient): {beta[2]:.4f}")
print(f"b3 (Assignment Score Coefficient): {beta[3]:.4f}")
```
